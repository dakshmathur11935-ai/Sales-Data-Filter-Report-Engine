```python
import numpy as np
import matplotlib.pyplot as plt
months = np.arange(1, 13)

month_names = np.array([
    "Jan", "Feb", "Mar", "Apr",
    "May", "Jun", "Jul", "Aug",
    "Sep", "Oct", "Nov", "Dec"
])

print(months)
print(month_names)
```

```python
stores = np.array([
    "Store 1", "Store 2", "Store 3", "Store 4", "Store 5",
    "Store 6", "Store 7", "Store 8", "Store 9", "Store 10"
])

print(stores)
```

```python
np.random.seed(42)

sales = np.random.randint(20000, 100001, size=(10, 12))

print(sales)
```

```python
print("Shape:", sales.shape)
print("Dimensions:", sales.ndim)
print("Total Elements:", sales.size)
print("Data Type:", sales.dtype)
```

```python
monthly_sales = np.sum(sales, axis=0)

for i in range(12):
    print(month_names[i], ":", monthly_sales[i])
```

```python
store_sales = np.sum(sales, axis=1)

for i in range(10):
    print(stores[i], ":", store_sales[i])
```

```python
q1 = sales[:, 0:3]
q2 = sales[:, 3:6]
q3 = sales[:, 6:9]
q4 = sales[:, 9:12]

print("Q1 Sales:", np.sum(q1))
print("Q2 Sales:", np.sum(q2))
print("Q3 Sales:", np.sum(q3))
print("Q4 Sales:", np.sum(q4))
```

```python
target = 60000

high_sales = sales > target

print(sales[high_sales])
```

```python
average_sales = np.mean(store_sales)

print("Average Store Sales:", average_sales)

below_average = store_sales < average_sales

print("Stores Below Average:")

for i in np.where(below_average)[0]:
    print(stores[i], ":", store_sales[i])
```

```python

condition1 = (sales > 60000) & (sales < 90000)

print("Sales between 60000 and 90000:")
print(sales[condition1])


condition2 = (sales < 30000) | (sales > 90000)

print("Sales below 30000 or above 90000:")
print(sales[condition2])
```

```python
selected_stores = sales[[1, 4, 7], :]

print(selected_stores)
```

```python
best_months = np.argsort(monthly_sales)[-3:][::-1]

for i in best_months:
    print(month_names[i], ":", monthly_sales[i])
```

```python
growth_targets = np.linspace(50000, 80000, 12)

print(growth_targets)
```

```python
target_achieved = monthly_sales >= growth_targets

for i in range(12):
    if target_achieved[i]:
        print(month_names[i], "Target Achieved")
    else:
        print(month_names[i], "Target Not Achieved")
```

```python
plt.figure(figsize=(10 , 5))

plt.bar(month_names, monthly_sales)

plt.title("Monthly Sales Report")
plt.xlabel("Months")
plt.ylabel("Total Sales")

plt.show()
```
